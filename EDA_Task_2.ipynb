```python
# Import the libraries we need
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/superstore_raw.csv', encoding='latin1')
```

```python
# Display the first 5 rows
df.head()
```

```python
# Find the total Sales for each Category
category_sales = df.groupby('Category')['Sales'].sum()

# Display the result
category_sales
```

```python
# Create a bar plot
category_sales.plot(kind='bar')

plt.title('Total Sales for Each Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

```python
# Create a box plot for Profit
plt.boxplot(df['Profit'])

plt.title('Profit Distribution')
plt.ylabel('Profit')
plt.show()
```

```python
# Create a scatter plot
plt.scatter(df['Discount'], df['Profit'])

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
```

```python
# Select numerical columns
numerical_data = df.select_dtypes(include='number')

# Create the correlation matrix
correlation_matrix = numerical_data.corr()

# Display the correlation matrix
correlation_matrix
```

```python
# Create a heatmap
sns.heatmap(correlation_matrix, annot=True)

plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
# Find total Profit for each Category
category_profit = df.groupby('Category')['Profit'].sum()

# Find the category with the highest total Profit
highest_profit_category = category_profit.idxmax()
highest_profit = category_profit.max()

print('Category with the highest total Profit:', highest_profit_category)
print('Total Profit:', highest_profit)
```
